```python
import pandas as pd

df = pd.read_csv('2018_Central_Park_Squirrel_Census_-_Squirrel_Data.csv')

```

```python


fur_colors = df['Primary Fur Color']
fur_colors.head(10)
```

```python

fur_colors = fur_colors.dropna()
```

```python

summary = (
    fur_colors
    .value_counts()                     # counts each unique value
    .reset_index(name='count')          # turns Series → DataFrame
    .sort_values('count', ascending=False)
    .rename(columns={'Primary Fur Color': 'Fur Color'})
)
```

```python
print(summary)
```
